```python
# Import Libraries
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
# Load dataset
df = pd.read_csv('https://raw.githubusercontent.com/niteen11/DataAnalyticsAcademy/master/Python/dataset_diabetes/diabetic_data.csv')
```

```python
# Preview data
df.head()
```

```python
# Understanding the data structure, value_counts() show how many patients fall into a category
df['readmitted'].value_counts()
df['time_in_hospital'].describe()
```

```python
# Data Cleaning, removing any records where hospital stay was 0 or invalid.
df = df[df['time_in_hospital'] > 0]
```

```python
# Group Comparison
df.groupby('readmitted')['time_in_hospital'].describe()
```

I used groupby() to split the data into groups based on readmission status. Then I applied describe() to calculate statistics for each group. This allows me to compare hospital stay patterns across different patient outcomes.

Data Visualization

```python
import matplotlib.pyplot as plt
import seaborn as sns

df.boxplot(column='time_in_hospital', by='readmitted')
plt.title('Hospital Stay by Readmission Status')
plt.suptitle('')
plt.xlabel('Readmitted')
plt.ylabel('Days in Hospital')
plt.show()
```

I created a boxplot to visually compare hospital stay across readmission groups. A boxplot shows the distribution of data, including the median, spread, and potential outliers. This helps make differences between groups easier to interpret.

Conclusion:

From the analysis, I observed that patients who were readmitted within 30 days generally had longer hospital stays compared to those who were not readmitted.

This suggests that longer hospital stays may be associated with more severe conditions, which could increase the likelihood of readmission.

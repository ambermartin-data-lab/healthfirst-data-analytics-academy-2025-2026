```python
# Rock Paper Scissors Game
# Two player interactive version

def get_player_choice(player_name):
    # def = defines a function (reusable block of code)
    # () = parentheses hold the parameter(s) the function receives
    # player_name = parameter (input variable) used inside the function
    # : = colon starts the indented code block for the function

    while True:
        # while = loop keyword
        # True = boolean value that is always true
        # while True: = an infinite loop (keeps running until we return or break)

        choice = input(f"{player_name}, enter rock, paper, or scissors: ").lower()
        # input() = asks the user for input
        # () = runs the function
        # "" = quotes create a string (text)
        # f"" = formatted string so {player_name} can be inserted into the text
        # .lower() = converts the string to lowercase for consistency

        if choice in ["rock", "paper", "scissors"]:
            # if = checks a condition
            # in = checks if a value exists in a list
            # [] = list of valid options
            # : = starts the indented block if the condition is True

            return choice
            # return = sends the value back to where the function was called

        else:
            # else: runs when the if condition is False
            print("Invalid input. Please type rock, paper, or scissors.")
            # print() = displays text to the screen


def determine_winner(player1, player2):
    # def = defines another function
    # player1, player2 = parameters (inputs) passed into the function
    # : = starts the indented code block for the function

    if player1 == player2:
        # if = checks a condition
        # == = “equal to” comparison operator
        return "It's a tie!"

    elif (player1 == "rock" and player2 == "scissors") or \
         (player1 == "scissors" and player2 == "paper") or \
         (player1 == "paper" and player2 == "rock"):
        # elif = “else if” (another condition if previous was False)
        # () = groups conditions together
        # and = BOTH parts must be True
        # or = allows multiple winning patterns
        # == = equal to
        return "Congratulations Player 1! You win!"

    else:
        # else: runs if none of the above conditions are True
        return "Congratulations Player 2! You win!"


print("Welcome to Rock Paper Scissors!")
# print() outputs text to the screen
# "" = string (text)

while True:
    # while True = repeats the whole game until break is hit

    player1_choice = get_player_choice("Player 1")
    # = assignment operator (stores value in a variable)

    player2_choice = get_player_choice("Player 2")

    result = determine_winner(player1_choice, player2_choice)
    # function call: determine_winner() uses () to pass arguments into the function

    print(result)

    play_again = input("Do you want to play again? (yes/no): ").lower()

    if play_again != "yes":
        # != = “not equal to”
        print("Game Over. Thanks for playing!")
        break
        # break = exits the loop immediately
```

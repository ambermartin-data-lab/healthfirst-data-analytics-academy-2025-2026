```python
import pandas as pd

df = pd.read_csv("https://raw.githubusercontent.com/CunyLaguardiaDataAnalytics/datasets/refs/heads/master/mtcars.csv")
```

```python
df.head()
```

```python
df.describe()
```

```python
df = df.rename(columns={"Unnamed: 0": "car"})
```

```python
df_subset = df[["mpg", "hp", "wt", "cyl"]]
```

```python
df_subset = df_subset.rename(columns={
    "mpg": "miles_per_gallon",
    "hp": "horsepower",
    "wt": "weight",
    "cyl": "cylinders"
})
```

```python
df_subset.mean()
```
